## Part 2: 🔐 Managing Secrets

**Never commit API keys to git!**

````python
# Secure secrets management

SECRETS_PATTERN = '''
# In your app.py
import streamlit as st
import os

# Method 1: Streamlit secrets (recommended for deployment)
def get_api_key():
    """Get API key from Streamlit secrets or user input."""
    # Try to get from Streamlit secrets first
    try:
        api_key = st.secrets["ANTHROPIC_API_KEY"]
        return api_key
    except (FileNotFoundError, KeyError):
        # Fall back to user input
        st.sidebar.warning("⚠️ API key not configured")
        api_key = st.sidebar.text_input(
            "Enter your Anthropic API key:",
            type="password",
            help="Get your key from console.anthropic.com"
        )
        if not api_key:
            st.error("Please provide an API key to continue")
            st.stop()
        return api_key

# Method 2: Environment variables (for local development)
from dotenv import load_dotenv
load_dotenv()  # Loads from .env file

api_key = os.getenv("ANTHROPIC_API_KEY")
'''

print('🔐 Secrets Management Pattern:')
print(SECRETS_PATTERN)
print('\n📝 For Hugging Face Spaces:')
print('  1. Go to Space settings')
print('  2. Add secrets in "Repository secrets" section')
print('  3. Access via st.secrets["KEY_NAME"]')
````

### Rollback Strategy

````python
# How to rollback if something breaks

ROLLBACK_COMMANDS = '''
# If deployment breaks:

# 1. Check logs
# Go to Space → Logs tab

# 2. Rollback to previous version
git log  # Find working commit hash
git revert <commit_hash>
git push

# OR hard reset (careful!)
git reset --hard <working_commit_hash>
git push --force

# 3. Space rebuilds with old version
'''

print('🔙 Rollback Strategy:')
print(ROLLBACK_COMMANDS)
````

### Exercise 2 (Intermediate): Deploy With Secrets

**Task**: Deploy an app that uses API keys

**Requirements**:
1. App uses Anthropic API
2. Secrets managed securely
3. Fallback to user input if no secret
4. Test both methods
5. Document setup in README

````python
import streamlit as st
import os
from google import genai

def initialize_gemini_client():
    """Reads API key securely from Streamlit secrets or falls back to user input."""
    api_key = None

    # 1. Attempt to load from Streamlit Secrets
    try:
        api_key = st.secrets["GEMINI_API_KEY"]
        st.sidebar.success("🔑 API Key loaded from Streamlit Secrets!")
    except (FileNotFoundError, KeyError, AttributeError):
        # 2. Fall back to local environment variables
        api_key = os.getenv("GEMINI_API_KEY")

    # 3. If still missing, ask the user via Streamlit Sidebar
    if not api_key:
        st.sidebar.warning("⚠️ GEMINI_API_KEY not configured in Secrets.")
        api_key = st.sidebar.text_input(
            "Enter your Gemini API Key:",
            type="password",
            help="Get your key from Google AI Studio."
        )
        if not api_key:
            st.info("👈 Please enter your Gemini API Key in the sidebar to continue.")
            st.stop()

    return genai.Client(api_key=api_key)

# --- App Interface Demo ---
st.title("🔐 Secure AI Assistant App")
client = initialize_gemini_client()

prompt = st.text_input("Ask anything:", "Explain deployment in one sentence.")
if st.button("Generate"):
    with st.spinner("Thinking..."):
        try:
            response = client.models.generate_content(
                model='gemini-2.5-flash-lite',
                contents=prompt,
            )
            st.write(response.text)
        except Exception as e:
            st.error(f"Error generating response: {e}")
````

### Exercise 3 (Intermediate): Performance Optimization

**Task**: Optimize a slow app

**Given**: App that loads data on every run

**Requirements**:
1. Add @st.cache_data where appropriate
2. Implement pagination
3. Add loading indicators
4. Measure before/after performance
5. Document improvements

````python
import streamlit as st
import pandas as pd
import time

# 1. Cache the heavy dataset load operation
@st.cache_data(show_spinner="Fetching and caching data...")
def get_heavy_dataset():
    # Simulating a slow network or database fetch
    time.sleep(3)
    data = pd.DataFrame({
        'ID': range(1, 101),
        'Product': [f'Product {i}' for i in range(1, 101)],
        'Sales': [i * 15.5 for i in range(1, 101)]
    })
    return data

st.title("⚡ Optimized & Cached App")

# Start timer to measure performance
start_time = time.time()

# Load data using cached function
data_df = get_heavy_dataset()

end_time = time.time()
st.success(f"⏱️ Data loaded in {end_time - start_time:.4f} seconds!")

# 2. Pagination for UI response optimization
items_per_page = 10
total_pages = len(data_df) // items_per_page

page = st.number_input("Page", min_value=1, max_value=total_pages, step=1)

start_idx = (page - 1) * items_per_page
end_idx = start_idx + items_per_page

st.subheader(f"Displaying items {start_idx + 1} to {end_idx}")
st.dataframe(data_df.iloc[start_idx:end_idx], use_container_width=True)
````

### Exercise 4 (Advanced): Complete Digital Twin Deployment

**Task**: Deploy your full Week 1-4 digital twin

**African Context**:
- Test on 3G connection

- Multiple language support

````python
import streamlit as st
import os
from google import genai

# Define African language options for localized support
LANGUAGES = {
    'English': 'Respond in English.',
    'Swahili': 'Respond in Swahili (Kiswahili).',
    'Yoruba': 'Respond in Yoruba.',
    'Zulu': 'Respond in Zulu (isiZulu).'
}

st.set_page_config(page_title="African Context Assistant", layout="centered")

st.title("🌍 Digital Twin Assistant (Low Bandwidth Optimized)")
st.caption("Optimized for 3G/Mobile connections. Keep requests concise to save bandwidth.")

# Secure client setup
try:
    api_key = st.secrets.get("GEMINI_API_KEY", os.getenv("GEMINI_API_KEY", ""))
except Exception:
    api_key = ""

if not api_key:
    api_key = st.sidebar.text_input("Enter Gemini API Key:", type="password")

selected_lang = st.sidebar.selectbox("Choose Language / Chagua Lugha:", list(LANGUAGES.keys()))

if not api_key:
    st.warning("Please configure your Gemini API key to start communicating.")
    st.stop()

client = genai.Client(api_key=api_key)

# Lightweight messaging history stored locally in session_state
if "messages" not in st.session_state:
    st.session_state.messages = []

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])

if prompt := st.chat_input("Ask your digital twin..."):
    st.session_state.messages.append({"role": "user", "content": prompt})
    with st.chat_message("user"):
        st.markdown(prompt)

    # Query LLM with language localization prompt injection
    with st.chat_message("assistant"):
        with st.spinner("..."): # Short spinner for reduced UI weight
            system_instruction = LANGUAGES[selected_lang]
            full_prompt = f"{system_instruction}\nUser Question: {prompt}"

            response = client.models.generate_content(
                model='gemini-2.5-flash-lite',
                contents=full_prompt,
            )
            answer = response.text
            st.markdown(answer)
            st.session_state.messages.append({"role": "assistant", "content": answer})
````
